```python
import numpy as np 
from random import randint


notas = []
for i in range(10000):
    notas.append(randint(0,100))

notas
```

```python
# divide por 100 e multiplica 2 
notas_convertidas = []
for nota in notas:   
    nova_nota = nota / 100 * 2
    notas_convertidas.append(nova_nota)

notas_convertidas
```

```python
# array
array_notas = np.array(notas) #criando o array de notas 
array_notas
```

```python
# é possivel realizar as operações sem fazer FOR/WHILE 
array_notas / 100 * 2 
```

```python
# Podemos realizar QUALQUER tipo de operação
array_notas < 1 

# TODO ARRAY DE VERDADEIRO/FALSO PODE SER USADO COMO FILTRO
array_notas[array_notas < 1]
```

```python
# EXERCICIO 1

precos = np.array([20.0, 45.50, 120.0, 85.0, 10.0])

precos_atualizados = precos * 1.15

print(precos_atualizados)
```

```python
# exercicio 2
 
mascara = precos_atualizados > 100 
precos_caros = precos_atualizados[mascara]

print(precos_caros)
```

```python
# EXERCICIO 3 
 
idades = np.array([15, 18, 22, 14, 30, 17, 45])

maiores_de_idade = idades[idades>=18]

print(maiores_de_idade)
```

```python
# EXERCICIO 4

notas = np.array([5.5, 7.0, 4.5, 9.5, 8.0])
notas_atualizadas = notas + 0.5

print(notas_atualizadas)
```

```python
# EXERCICIO 5 

estoque = np.array([10, -2, 5, 0, -1, 20])
estoque_valido = estoque[estoque>=0]

print(estoque_valido)
```

```python
# EXERCICIO 6 

vendas = np.array([150, 200, 100, 350, 400])

total = vendas.sum()
media = vendas.mean()

print("total:", total)
print("media:", media)
```

```python
# EXERCICIO 7

idade = np.array([12, 25, 30, 45, 15, 60])

filtro = idades[(idades > 20 ) & (idades < 40)]

print(filtro)
```

```python
# EXERCICIO 8

temp = np.array([25, -999, 30, 22, -999])
temp_corrigida = np.where(temp == -999, 0, temp)
print(temp_corrigida)
```

```python
# EXERCICIO 9 

valores = np.array([1.5, 2.0, np.nan, 4.5, np.nan])

mascara = ~np.isnan(valores)

valores_validos = valores[mascara]

print(valores_validos)
```
